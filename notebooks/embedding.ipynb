```python
sen = "This is a not test sent, with some. Didn't  doesnt! 324234 4 3 2 44"
```

```python
tok = nltk.word_tokenize(sen)
tok
```

```python
for t in tok:
    try:
        wv[t]
    except KeyError as e:
        print(e)
```

```python
from art.preprocess import to_tokens
```

```python
tok, oov, oov_rate = to_tokens(sen, wv)
tok
```

```python
from art.embedding import add_embed
import re
```

```python
cols = ["observation_1", "observation_2", "hypothesis_1", "hypothesis_2"]


def has_empty_field(x):
    return any(not re.search(r"\w", x[c]) for c in cols)


ds["train"] = ds["train"].filter(lambda x: not has_empty_field(x))
```

```python
ds = add_embed(ds, wv)
```
