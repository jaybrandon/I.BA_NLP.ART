```python
from art.models import load_w2v
from art.data import get_dataset
from art.embedding import add_embed
from torch.utils.data import DataLoader
import torch
```

```python
wv = load_w2v()
```

```python
ds = get_dataset()

ds = add_embed(ds, wv)
```

```python
ds
```

```python
ds["train"][0]
```

```python
train_ds = ds["train"].with_format("torch", ["embed", "label"])
```

```python
train_ds
```

```python
train_loader = DataLoader(train_ds, batch_size=64, shuffle=True)
```

```python
batch = next(iter(train_loader))
```
